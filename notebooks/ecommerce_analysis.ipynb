## 4. Feature Engineering

```python
df["Year"] = df["Order_Date"].dt.year
```

```python
df["Month"] = df["Order_Date"].dt.month
```

```python
df["Month_Name"] = df["Order_Date"].dt.month_name()
```

```python
df["Quarter"] = df["Order_Date"].dt.quarter
```

```python
df["Quarter"] = "Q" + df["Quarter"].astype(str)
```

```python
df["Year_Month"] = df["Order_Date"].dt.to_period("M")
```

```python
df[[
    "Order_Date",
    "Year",
    "Month",
    "Month_Name",
    "Quarter",
    "Year_Month"
]].head()
```

```python
df["Profit_Margin"] = (df["Profit"] / df["Sales"]) * 100
```

```python
df[["Sales", "Profit", "Profit_Margin"]].head()
```

```python
df["Profit_Margin"] = (df["Profit"] / df["Sales"]) * 100

df["Profit_Margin_Display"] = df["Profit_Margin"].map("{:.2f}%".format)

df[["Sales", "Profit", "Profit_Margin_Display"]].head()
```

## 5. Exploratory Data Analysis (EDA)

```python
df.describe()
```

```python
total_revenue = df["Sales"].sum()

print(f"Total Revenue: ₹{total_revenue:,.2f}")
```

```python
total_profit = df["Profit"].sum()

print(f"Total Profit: ₹{total_profit:,.2f}")
```

```python
total_orders = df["Order_ID"].nunique()

print(f"Total Orders: {total_orders:,}")
```

```python
total_customers = df["Customer_ID"].nunique()

print(f"Total Customers: {total_customers:,}")
```

```python
total_quantity = df["Quantity"].sum()

print(f"Total Quantity Sold: {total_quantity:,}")
```

```python
aov = total_revenue / total_orders

print(f"Average Order Value: ₹{aov:,.2f}")
```

```python
overall_profit_margin = (
    total_profit / total_revenue
) * 100

print(f"Overall Profit Margin: {overall_profit_margin:.2f}%")
```

```python
print("===== E-COMMERCE KPI SUMMARY =====")
print(f"Total Revenue       : ₹{total_revenue:,.2f}")
print(f"Total Profit        : ₹{total_profit:,.2f}")
print(f"Total Orders        : {total_orders:,}")
print(f"Total Customers     : {total_customers:,}")
print(f"Total Quantity Sold : {total_quantity:,}")
print(f"Average Order Value : ₹{aov:,.2f}")
print(f"Profit Margin       : {overall_profit_margin:.2f}%")
```

```python
monthly_sales = (
    df.groupby("Year_Month")["Sales"]
      .sum()
      .reset_index()
)

monthly_sales.head()
```

```python
monthly_sales["Year_Month"] = monthly_sales["Year_Month"].astype(str)
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales["Year_Month"],
    monthly_sales["Sales"],
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

plt.show()
```
